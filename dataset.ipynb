# STEP 1: IMPORT LIBRARIES

```python
import pandas as pd
import numpy as np
import random
from datetime import datetime, timedelta
```

# STEP 2: SET A SEED (for reproducibility)

```python
random.seed(42)
np.random.seed(42)
```

# STEP 3: DEFINE POSSIBLE VALUES

```python
categories = ['Login Issue', 'Malware Alert', 'Access Request','Slow Performance', 'Phishing Email', 
              'Data Export']
```

generally for real time IT Tickets these are going to be the possible values listed in the ticket

"These lists represent the valid values for ticket categories and priorities in an IT support system. They allow to generate realistic synthetic data using random selection, maintain data consistency, and later perform analyses such as ticket distribution, priority trends, and security incident tracking."

```python
priorities = ['Low', 'Medium', 'High', 'Critical']
```

# STEP 4: CREATE 1000 ROWS OF DATA

```python
rows = []

for i in range(1000):
     # Pick a random category
    cat = random.choice(categories)
    
    # Pick a random priority
    pri = random.choice(priorities)
    
    # Random resolution time (1 to 72 hours)
    res_time = random.randint(1, 72)
    
    # Did this user have the same issue before? (30% chance = Yes)
    repeated = random.random() < 0.3  # True or False
    
    # Was ticket raised after 6pm or before 8am? (25% chance)
    after_hrs = random.random() < 0.25  # True or False
    
    breach_prob = 0.05  # base: only 5% are breaches (realistic!)
    
    if cat in ['Malware Alert', 'Phishing Email']:
        breach_prob += 0.30  # these categories are riskier
    if pri in ['High', 'Critical']:
        breach_prob += 0.15
    if after_hrs:
        breach_prob += 0.10  # hackers often attack at night
    if repeated:
        breach_prob += 0.10  # repeated = suspicious pattern
    
    is_breach = random.random() < breach_prob  # True/False
    
     # Create a random date in 2023
    start_date = datetime(2023, 1, 1)
    rand_days = random.randint(0, 364)
    ticket_date = start_date + timedelta(days=rand_days)
    
    rows.append({
        'ticket_id': f'T{i+1:04d}',     # T0001, T0002...
        'date': ticket_date.strftime('%Y-%m-%d'),
        'category': cat,
        'priority': pri,
        'resolution_time': res_time,
        'repeated_issue': repeated,
        'after_hours': after_hrs,
        'is_breach': is_breach
    })
```

in general we consider breach:-
# - Malware + High/Critical = more likely breach
# - Phishing + after hours = more likely breach
# - Repeated issue + Critical = suspicious

"I created rule-based business logic. Every ticket starts with a 5% breach probability. Risk factors such as malware alerts, phishing emails, high priority, after-hours activity, and repeated issues increase the probability. A random draw is then used to determine the final breach outcome. This simulates how multiple risk indicators contribute to security incidents in a real IT environment."

```python
# Convert list of rows into a DataFrame (table)
df = pd.DataFrame(rows)
```
